# Saved-run methods and input report
CPU → Run all. Reads saved counts/onset tables; no raw EEG extraction. Saves small CSV/PNG reports under each existing run’s analysis folder. Original EEG and old results stay intact. New label-policy availability is not newly accepted EEG.

```python
from google.colab import drive
from pathlib import Path
import subprocess, sys, os
from IPython.display import display, Markdown, Image
import pandas as pd
drive.mount('/content/drive')
THESIS=Path('/content/drive/MyDrive/Thesis')
REPO=Path('/content/word-sentiment-eeg-methods')
if not REPO.exists():
    subprocess.run(['git','clone','--depth','1','https://github.com/parmisbathayan/word-sentiment-eeg.git',str(REPO)],check=True)
else:
    subprocess.run(['git','-C',str(REPO),'pull','--ff-only'],check=True)
print('Code:', subprocess.check_output(['git','-C',str(REPO),'rev-parse','--short','HEAD'],text=True))
```

## ZUCO: exclusions and input availability

```python
RUN=THESIS/'Results/word_sentiment_eeg/v4_word_erp_compact'
subprocess.run([sys.executable,str(REPO/'scripts/report_exclusion_flow.py'),'--run-dir',str(RUN),'--dataset','zuco'],check=True)
FLOW=RUN/'analysis/exclusion_flow_v1'
display(Markdown((FLOW/'REPORT.md').read_text()))
display(Image(filename=str(FLOW/'exclusion_flow.png')))
display(pd.read_csv(FLOW/'per_reader_class_counts.csv'))
LABELS=REPO/'annotations/word_labels_three_level.csv'
if LABELS.is_file():
    subprocess.run([sys.executable,str(REPO/'scripts/report_input_inventory.py'),'--run-dir',str(RUN),'--dataset','zuco','--annotations',str(LABELS)],check=True)
    INV=RUN/'analysis/three_level_input_inventory'
    display(pd.read_csv(INV/'input_counts.csv'))
    display(pd.read_csv(INV/'per_reader_input_counts.csv'))
else:
    print('Three-level input inventory needs the private new label file:',LABELS)
```
